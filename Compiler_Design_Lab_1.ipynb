1.Write a python Program to implement  a simple table with operations to insert symbols,search for a symbol and display the content of the table.

```python
# Symbol Table Implementation

symbol_table = {}

def insert_symbol():
    name = input("Enter symbol name: ")

    if name in symbol_table:
        print("Symbol already exists.\n")
        return

    dtype = input("Enter data type: ")
    address = input("Enter memory address: ")

    symbol_table[name] = {
        "Type": dtype,
        "Address": address
    }

    print("Symbol inserted successfully.\n")


def search_symbol():
    name = input("Enter symbol to search: ")

    if name in symbol_table:
        print("\nSymbol Found:")
        print("Type    :", symbol_table[name]["Type"])
        print("Address :", symbol_table[name]["Address"])
    else:
        print("Symbol not found.")
    print()


def display_table():
    if not symbol_table:
        print("\nSymbol table is empty.\n")
        return

    print("\n------ SYMBOL TABLE ------")
    print("{:<15} {:<15} {:<15}".format("Symbol", "Type", "Address"))

    for key, value in symbol_table.items():
        print("{:<15} {:<15} {:<15}".format(
            key,
            value["Type"],
            value["Address"]
        ))
    print()


while True:
    print("===== SYMBOL TABLE MENU =====")
    print("1. Insert")
    print("2. Search")
    print("3. Display")
    print("4. Exit")

    try:
        choice = int(input("Enter your choice: "))
    except ValueError:
        print("Please enter a valid number.\n")
        continue

    if choice == 1:
        insert_symbol()
    elif choice == 2:
        search_symbol()
    elif choice == 3:
        display_table()
    elif choice == 4:
        print("Exiting...")
        break
    else:
        print("Invalid choice. Please try again.\n")
```

2. Write a python program to implement all basic symbols table operation including
modification and deletion.

```python
symbol_table = {}
def insert():
    symbol = input("Enter Symbol Name: ")
    datatype = input("Enter Data Type: ")

    if symbol in symbol_table:
        print("Symbol already exists.")
    else:
        symbol_table[symbol] = datatype
        print("Symbol inserted successfully.")

# Function to search a symbol
def search():
    symbol = input("Enter Symbol Name to Search: ")

    if symbol in symbol_table:
        print(f"Found -> {symbol} : {symbol_table[symbol]}")
    else:
        print("Symbol not found.")

# Function to modify a symbol
def modify():
    symbol = input("Enter Symbol Name to Modify: ")

    if symbol in symbol_table:
        new_datatype = input("Enter New Data Type: ")
        symbol_table[symbol] = new_datatype
        print("Symbol modified successfully.")
    else:
        print("Symbol not found.")

# Function to delete a symbol
def delete():
    symbol = input("Enter Symbol Name to Delete: ")

    if symbol in symbol_table:
        del symbol_table[symbol]
        print("Symbol deleted successfully.")
    else:
        print("Symbol not found.")

# Function to display symbol table
def display():
    if len(symbol_table) == 0:
        print("Symbol Table is Empty.")
    else:
        print("\nSymbol Table")
        print("----------------------")
        print("Symbol\tData Type")
        print("----------------------")
        for symbol, datatype in symbol_table.items():
            print(f"{symbol}\t{datatype}")
        print("----------------------")

# Main Menu
while True:
    print("\n===== SYMBOL TABLE MENU =====")
    print("1. Insert")
    print("2. Search")
    print("3. Modify")
    print("4. Delete")
    print("5. Display")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == '1':
        insert()
    elif choice == '2':
        search()
    elif choice == '3':
        modify()
    elif choice == '4':
        delete()
    elif choice == '5':
        display()
    elif choice == '6':
        print("Program Exited.")
        break
    else:
        print("Invalid Choice! Try Again.")
```
