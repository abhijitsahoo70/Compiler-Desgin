write a python program to read assemble label and their address and generate a symbol table.

```python

symbol_table = {}
n = int(input("Enter the number of labels: "))
for i in range(n):
    label = input(f"Enter label {i + 1}: ")
    address = input(f"Enter address of {label}: ")
    symbol_table[label] = address
print("\nSYMBOL TABLE")
print("-" * 30)
print("{:<15}{}".format("Label", "Address"))
print("-" * 30)

for label, address in symbol_table.items():
    print("{:<15}{}".format(label, address))
```
