1. WAP to desgin a DFA that reccocnize valid identifier(C keyword).

```python
identifier = input("Enter an identifier: ")

state = "q0"

for ch in identifier:
    if state == "q0":
        if ch.isalpha() or ch == "_":
            state = "q1"
        else:
            state = "q2"

    elif state == "q1":
        if ch.isalpha() or ch.isdigit() or ch == "_":
            state = "q1"
        else:
            state = "q2"

    elif state == "q2":
        break

if state == "q1":
    print("Accepted(Valid Identifier)")
else:
    print("Rejected(Invalid Identifier)")
```

2. WAP in python to implement a DFA that recognizes integer and floating point constant

```python
number = input("Enter a number: ")

state = "q0"

for ch in number:
    if state == "q0":
        if ch.isdigit():
            state = "q1"
        else:
            state = "q4"

    elif state == "q1":
        if ch.isdigit():
            state = "q1"
        elif ch == ".":
            state = "q2"
        else:
            state = "q4"

    elif state == "q2":
        if ch.isdigit():
            state = "q3"
        else:
            state = "q4"

    elif state == "q3":
        if ch.isdigit():
            state = "q3"
        else:
            state = "q4"

    elif state == "q4":
        break

if state == "q1" or state == "q3":
    print("Accepted")
else:
    print("Rejected")
```
