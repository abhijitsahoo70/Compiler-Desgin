1.write a program lexical analyser(scaner) to identify keywords, operator,constant,delemeters,and special symbol

```
import keyword
import re

# Define operators, delimiters, and special symbols
operators = ['+', '-', '*', '/', '%', '=', '==', '!=', '<', '>', '<=', '>=']
delimiters = ['(', ')', '{', '}', '[', ']', ';', ',', ':']
special_symbols = ['#', '@', '$', '&']

# Input from user
code = input("Enter a statement: ")

# Split the input into tokens
tokens = re.findall(r'\w+|==|!=|<=|>=|[^\s\w]', code)

print("\nLexical Analysis:\n")

for token in tokens:
    if keyword.iskeyword(token):
        print(f"{token} --> Keyword")
    elif token in operators:
        print(f"{token} --> Operator")
    elif token in delimiters:
        print(f"{token} --> Delimiter")
    elif token in special_symbols:
        print(f"{token} --> Special Symbol")
    elif re.fullmatch(r'\d+(\.\d+)?', token):
        print(f"{token} --> Constant")
    elif re.fullmatch(r'[A-Za-z_]\w*', token):
        print(f"{token} --> Identifier")
    else:
        print(f"{token} --> Unknown")
```

2.write a python program that counts numbre keywords,identifiers,operators,Delimiters,literers, in a given C source program

```
import re

# C keywords
keywords = {
    "auto", "break", "case", "char", "const", "continue", "default",
    "do", "double", "else", "enum", "extern", "float", "for", "goto",
    "if", "int", "long", "register", "return", "short", "signed",
    "sizeof", "static", "struct", "switch", "typedef", "union",
    "unsigned", "void", "volatile", "while"
}

# Operators and delimiters
operators = ['+', '-', '*', '/', '%', '=', '==', '!=', '<', '>', '<=', '>=',
             '&&', '||', '!', '++', '--']

delimiters = ['(', ')', '{', '}', '[', ']', ';', ',', ':']

# Read C program
code = input("Enter C source program: ")

# Tokenize the input
tokens = re.findall(r'"[^"]*"|\d+\.\d+|\d+|==|!=|<=|>=|\+\+|--|&&|\|\||\w+|[^\s\w]', code)

# Counters
keyword_count = 0
identifier_count = 0
operator_count = 0
delimiter_count = 0
literal_count = 0

# Analyze tokens
for token in tokens:
    if token in keywords:
        keyword_count += 1
    elif token in operators:
        operator_count += 1
    elif token in delimiters:
        delimiter_count += 1
    elif re.fullmatch(r'\d+(\.\d+)?', token) or re.fullmatch(r'"[^"]*"', token):
        literal_count += 1
    elif re.fullmatch(r'[A-Za-z_][A-Za-z0-9_]*', token):
        identifier_count += 1

# Display results
print("\nToken Counts:")
print("Keywords    :", keyword_count)
print("Identifiers :", identifier_count)
print("Operators   :", operator_count)
print("Delimiters  :", delimiter_count)
print("Literals    :", literal_count)
```
