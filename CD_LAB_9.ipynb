```python
from collections import defaultdict

def F(x):
    if x not in G:return{x}
    r=set()
    for p in G[x]:
        if p==['e']:r.add('e');continue
        for y in p:
            f=F(y);r|=f-{'e'}
            if'e'not in f:break
        else:r.add('e')
    return r

print("LL(1) PARSING TABLE GENERATOR\n------------------------------")
G={}
for _ in range(int(input("Enter number of productions: "))):
    a,b=input("Enter production: ").split("->");G[a]=[list(x) for x in b.split("|")]

S=next(iter(G));FIRST={a:F(a) for a in G};FOLLOW=defaultdict(set);FOLLOW[S]={'$'}

for _ in G:
    for A in G:
        for p in G[A]:
            for i,B in enumerate(p):
                if B in G:
                    f=F(p[i+1]) if i+1<len(p) else {'e'}
                    FOLLOW[B]|=f-{'e'}
                    if'e'in f:FOLLOW[B]|=FOLLOW[A]

T=sorted({x for p in G.values() for q in p for x in q if x not in G and x!='e'}|{'$'})
M={}
for A in G:
    for p in G[A]:
        f=set()
        for x in p:
            z=F(x);f|=z-{'e'}
            if'e'not in z:break
        else:f.add('e')
        for t in f-{'e'}:M[A,t]=p

print("\nFIRST SETS\n----------")
for A in G:print("FIRST("+A+") =",FIRST[A])
print("\nFOLLOW SETS\n-----------")
for A in G:print("FOLLOW("+A+") =",FOLLOW[A])
print("\nLL(1) PARSING TABLE\n-------------------")
print("NT/T".ljust(10),*[t.center(15) for t in T])
print("-"*(10+15*len(T)))
for A in G:print(A.ljust(10),*( (A+"->"+''.join(M[A,t])).center(15) if (A,t) in M else "-".center(15) for t in T))
```
